```python
# Cargar librerías
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv')
```

```python
print(df.head())
```

```python
print(df.info())
```

```python
promedio_edad = df['Age'].mean()
print(f"El promedio de edad de los pasajeros es: {promedio_edad:.2f} años.\n")
```

```python
df_filtrado = df[(df['Pclass'] == 1) & (df['Fare'] > 100)]
print(f"Cantidad de pasajeros en 1ra clase con tarifa mayor a $100: {len(df_filtrado)} pasajeros.\n")
```

```python
print("--- Valores nulos por columna ---")
print(df.isnull().sum())
print("\n")
```

```python
mediana_edad = df['Age'].median()
df['Age'] = df['Age'].fillna(mediana_edad)
```

```python
print("--- Valores nulos por columna ---")
print(df.isnull().sum())
print("\n")
```

```python
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
print(df[['Name', 'SibSp', 'Parch', 'FamilySize']].head(3))
```

```python
tasa_supervivencia_genero = df.groupby('Sex')['Survived'].mean() * 100
print("--- Tasa de supervivencia por género (%) ---")
print(tasa_supervivencia_genero)
print("\n")
```

```python
mean = df.groupby('Pclass')['Fare'].mean()
min = df.groupby('Pclass')['Fare'].min()
max = df.groupby('Pclass')['Fare'].max()
print(mean)
print(min)
print(max)
```
