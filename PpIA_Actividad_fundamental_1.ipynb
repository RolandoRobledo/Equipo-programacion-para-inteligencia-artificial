#### **Ejercicio 1: Evaluador de Rendimiento de Modelos**


 **Lógica implementada:** Para este ejercicio usamos una función llamada evaluar_modelo() que le pide al usuario la exactitud del modelo con input(). Como input() devuelve texto, lo convertimos a float para poder compararlo con números. También agregamos un while con try/except para que, si el usuario escribe algo que no es número o un valor fuera del rango de 0.0 a 1.0, el programa le vuelva a preguntar en lugar de marcar error. Después usamos if, elif y else para revisar en qué rango cae la exactitud: en los rangos de en medio utilicé el operador and para que el valor cumpla los dos límites a la vez. Al final la función imprime la categoría (Excelente, Aceptable, Requiere Ajuste o Crítico / No apto) y también la regresa con return.


---


```python
def evaluar_modelo():
    # Viendo que el valor este entre 0.0 y 1.0
    while True:
        try:
            exactitud = float(input("Ingresa la exactitud del modelo (0.0 a 1.0): "))
            if 0.0 <= exactitud <= 1.0:
                break
            print("El valor debe estar entre 0.0 y 1.0.")
        except ValueError:
            print("Entrada inválida. Escribe un número, por ejemplo 0.85.")

    # Clasificación con los operadores lógicos
    if exactitud >= 0.90:
        categoria = "Excelente"
    elif exactitud >= 0.75 and exactitud < 0.90:
        categoria = "Aceptable"
    elif exactitud >= 0.50 and exactitud < 0.75:
        categoria = "Requiere Ajuste"
    else:
        categoria = "Crítico / No apto"

    print(f"Exactitud: {exactitud:.2f} -> Categoría: {categoria}")
    return categoria

evaluar_modelo()
```

#### **Ejercicio 2: Procesamiento de Hiperparámetros y Lotes**


**Logica Implementada:** Primero definimos CONFIGURACION_BASE como una tupla global con los valores (512, 10), que representan la dimensión de entrada y de salida de una capa. Usamos una tupla porque no se puede modificar, que es lo que pedía la actividad. Para probar la función generamos una lista de 5 tasas de aprendizaje aleatorias con el módulo random, usando random.uniform() y round() para limitar los decimales. La función procesar_tasas_aprendizaje() recorre la lista con un ciclo for, multiplica cada tasa por un factor de escala de 2.0 y guarda el resultado con append() en una lista nueva, para no alterar la original. Al terminar, imprime y regresa la lista escalada junto con la tupla base.

```python
import random
# Tupla global: (dimensión de entrada, dimensión de salida). No se modifica.
CONFIGURACION_BASE = (512, 10)

def procesar_tasas_aprendizaje(tasas, factor_escala=2.0):
    tasas_escaladas=[]
    for tasa in tasas:
        tasas_escaladas.append(round(tasa*factor_escala,4))

    print("Tasas originales:  ", tasas)
    print("Tasas escaladas:   ", tasas_escaladas)
    print("Configuración base:", CONFIGURACION_BASE)
    return tasas_escaladas, CONFIGURACION_BASE

# Lista aleatoria de 5
lista_tasas=[]
for i in range(5):
    lista_tasas.append(round(random.uniform(0.0001, 0.1), 4))
resultado = procesar_tasas_aprendizaje(lista_tasas)
```

#### **Ejercicio 3: Simulador de Convergencia por Épocas**


**Lógica implementada:** Las epocas son la cantidad de veces que una IA repite su proceso de aprendizaje, entonces en simular_entrenamiento() pedimos al usuario el número máximo de épocas con input(), lo convertimos a entero y validamos que sea mayor que cero. Después usamos un ciclo for con range(1, max_epocas + 1) para recorrer las épocas desde la 1 hasta el límite. En cada época calculamos una pérdida simulada con la fórmula 2.0 / epoca, de modo que el valor disminuye conforme avanzan las épocas, como pasa en un entrenamiento real. Dentro del ciclo colocamos un if que compara la pérdida con el umbral de 0.05; si es menor, se imprime que el modelo convergió anticipadamente y se usa break para detener el ciclo. Si el ciclo termina sin llegar al umbral, el else del for avisa que se alcanzó el máximo de épocas sin converger. Por ejemplo, con 50 épocas el programa se detiene en la época 41, donde la pérdida es de aproximadamente 0.049.

```python
def simular_entrenamiento():
    # Validación de la entrada
    while True:
        try:
            max_epocas = int(input("Número máximo de épocas a simular: "))
            if max_epocas > 0:
                break
            print("Debe ser un entero mayor que 0.")
        except ValueError:
            print("Entrada inválida. Escribe un número entero.")

    UMBRAL = 0.05
    for epoca in range(1, max_epocas + 1):
        # Pérdida simulade
        loss = 2.0 / epoca
        print(f"Época {epoca:>3}/{max_epocas} - loss: {loss:.4f}")

        if loss < UMBRAL:
            print(f"El modelo ha convergido anticipadamente en la época {epoca} "
                  f"(loss {loss:.4f} < {UMBRAL}).")
            break
    else:
        print("Se alcanzó el máximo de épocas sin converger.")

simular_entrenamiento()
```
